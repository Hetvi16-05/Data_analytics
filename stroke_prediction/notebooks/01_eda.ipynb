# 01 - Exploratory Data Analysis (EDA)
In this notebook, we explore the stroke dataset, check for missing values, and analyze distributions.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
df = pd.read_csv('../data/raw/healthcare-dataset-stroke-data.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

## Missing Values Analysis


```python
df.isnull().sum()
```

BMI has missing values (we will handle these in the preprocessing notebook).


## Target Variable Distribution


```python
sns.countplot(x='stroke', data=df)
plt.title('Stroke Distribution (Class Imbalance)')
plt.show()
```

## Feature Analysis


```python
sns.histplot(df['age'], bins=30, kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.boxplot(x='stroke', y='age', data=df)
plt.title('Stroke vs Age')
plt.show()
```
