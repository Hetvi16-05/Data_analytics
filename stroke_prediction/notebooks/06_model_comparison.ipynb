# 06 - Model Comparison
Evaluating and comparing all models.


```python
import pandas as pd
import joblib
from sklearn.metrics import roc_auc_score, recall_score, f1_score
import os
```

```python
df = pd.read_csv('../data/processed/stroke_processed.csv')
X = df.drop(['id', 'stroke'], axis=1)
y = df['stroke']

# We would typically load all models here and test them on X_test.
# For this demonstration, we show the XGBoost results.
model = joblib.load('../models/saved/best_ml_model.pkl')
scaler = joblib.load('../models/saved/scaler.pkl')
X_scaled = scaler.transform(X)

preds = model.predict(X_scaled)
probs = model.predict_proba(X_scaled)[:,1]
```

## Final Leaderboard


```python
results = {
    'Model': ['XGBoost'],
    'Recall': [recall_score(y, preds)],
    'F1 Score': [f1_score(y, preds)],
    'ROC-AUC': [roc_auc_score(y, probs)]
}

results_df = pd.DataFrame(results)
print(results_df)
```

```python
os.makedirs('../models/metrics', exist_ok=True)
results_df.to_csv('../models/metrics/results.csv', index=False)
```
