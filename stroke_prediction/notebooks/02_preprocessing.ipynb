# 02 - Data Preprocessing & Feature Engineering
Handling missing values, creating new features, and encoding categoricals.


```python
import pandas as pd
import numpy as np
import os
```

```python
df = pd.read_csv('../data/raw/healthcare-dataset-stroke-data.csv')
```

## Handling Missing Values & Types


```python
df['bmi'] = pd.to_numeric(df['bmi'].replace('N/A', np.nan), errors='coerce')
df['bmi'] = df['bmi'].fillna(df['bmi'].median())
df = df[df['gender'] != 'Other'] # Drop 'Other' gender (only 1 row)
df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['age'] = df['age'].astype(int)
```

## Encoding Categorical Variables


```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
binary_cols = ['ever_married', 'Residence_type', 'gender']
for col in binary_cols:
    df[col] = le.fit_transform(df[col])

df = pd.get_dummies(df, columns=['work_type', 'smoking_status'], drop_first=True)
```

## Save Processed Data


```python
os.makedirs('../data/processed', exist_ok=True)
df.to_csv('../data/processed/stroke_processed.csv', index=False)
df.head()
```
