# 05 - SHAP Explainability
Understanding feature importance using SHAP values on our best model.


```python
import pandas as pd
import shap
import joblib
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/processed/stroke_processed.csv')
X = df.drop(['id', 'stroke'], axis=1)

model = joblib.load('../models/saved/best_ml_model.pkl')
scaler = joblib.load('../models/saved/scaler.pkl')
X_scaled = scaler.transform(X)
```

## Calculate SHAP Values


```python
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X_scaled)
```

## SHAP Summary Plot


```python
shap.summary_plot(shap_values, X, plot_type='bar')
```
